```python
def goal_compatibility(student_data, goal):
    
    profile = goal_profiles.loc[goal]
    
    # Important capability weights
    weights = {
        "Math_Score": 0.20,
        "Logical_Ability": 0.20,
        "Aptitude_Score": 0.20,
        "Programming_Skill": 0.15,
        "Academic_Performance": 0.10,
        "Communication_Skill": 0.05,
        "Soft_Skills_Score": 0.05,
        "Networking_Score": 0.05
    }
    
    score = 0
    total_weight = 0
    
    for column, weight in weights.items():
        
        student_value = student_data[column]
        goal_value = profile[column]
        
        # Convert difference into similarity
        if column in ["Programming_Skill", "Communication_Skill", "Logical_Ability"]:
            max_value = 5
        elif column in ["Soft_Skills_Score", "Networking_Score"]:
            max_value = 10
        else:
            max_value = 100
        
        similarity = 1 - abs(student_value - goal_value) / max_value
        
        similarity = max(0, similarity)
        
        score += similarity * weight
        total_weight += weight
    
    return score / total_weight
```

```python
def generate_final_recommendation(student_df):

    # -----------------------------
    # 1. Model prediction
    # -----------------------------
    processed = preprocessor.transform(student_df)

    probabilities = model.predict_proba(processed)[0]

    result = pd.DataFrame({
        "Career": model.classes_,
        "Model_Confidence": probabilities * 100
    })

    # -----------------------------
    # 2. Capability match
    # -----------------------------
    student = student_df.iloc[0]
    
    capability_scores = []

    for career in model.classes_:

        career_data = df[
            df["Recommended_Career"] == career
        ]

        score = 0
        count = 0

        for col in capability_columns:

            career_mean = career_data[col].mean()

            data_range = df[col].max() - df[col].min()

            if data_range > 0:

                similarity = (
                    1 -
                    abs(student[col] - career_mean)
                    / data_range
                )

                similarity = max(0, similarity)

                score += similarity
                count += 1

        capability_match = (
            score / count
        ) * 100

        capability_scores.append(capability_match)

    result["Capability_Match"] = capability_scores

    # -----------------------------
    # 3. Goal alignment
    # -----------------------------
    goal = student["Career_Goal"]

    goal_counts = pd.crosstab(
        df["Career_Goal"],
        df["Recommended_Career"]
    )

    if goal in goal_counts.index:

        goal_distribution = (
            goal_counts.loc[goal]
            / goal_counts.loc[goal].sum()
        ) * 100

    else:

        goal_distribution = pd.Series(
            0,
            index=model.classes_
        )

    result["Goal_Alignment"] = [
        goal_distribution.get(career, 0)
        for career in result["Career"]
    ]

    # -----------------------------
    # 4. Normalize model score
    # -----------------------------
    result["Model_Score"] = (
        result["Model_Confidence"]
        / result["Model_Confidence"].max()
    ) * 100

    # -----------------------------
    # 5. Final score
    # -----------------------------
    result["Final_Score"] = (
        0.40 * result["Model_Score"] +
        0.35 * result["Capability_Match"] +
        0.25 * result["Goal_Alignment"]
    )

    # -----------------------------
    # 6. Sort
    # -----------------------------
    result = result.sort_values(
        "Final_Score",
        ascending=False
    ).reset_index(drop=True)

    return result.head(5)
```

```python
# ==========================================
# USER INPUT → MODEL FEATURES
# ==========================================

def build_user_features(
    aptitude,
    interest,
    personality,
    academic_performance,
    skills,
    career_goal
):
    
    # Base profile
    student = {
        "Aptitude_Score": aptitude,
        "Interest": interest,
        "Personality_Type": personality,
        "Academic_Performance": academic_performance,
        "Skills": skills,
        "Career_Goal": career_goal
    }

    # Derived features
    student["Overall_Aptitude"] = aptitude

    student["Academic_Strength"] = academic_performance

    # Initial skill-based estimates
    # These will later be refined using the selected skills
    student["Programming_Skill"] = 3
    student["Communication_Skill"] = 3
    student["Soft_Skills_Score"] = 7

    student["Technical_Skill_Strength"] = (
        student["Programming_Skill"] / 5 * 100
    )

    student["Communication_Strength"] = (
        student["Communication_Skill"] / 5 * 100
    )

    student["Career_Readiness_Score"] = (
        academic_performance * 0.5 +
        aptitude * 0.5
    )

    return student


print("User input mapping function created successfully!")
```

```python
# ==========================================
# SKILLS → MODEL SCORES
# ==========================================

def calculate_skill_scores(skills):

    skills_text = skills.lower()

    technical_keywords = [
        "python", "sql", "machine learning", "deep learning",
        "data science", "programming", "java", "c++",
        "cloud", "aws", "azure", "devops", "linux",
        "cyber security", "cybersecurity", "web development"
    ]

    communication_keywords = [
        "communication", "presentation", "public speaking",
        "negotiation", "sales", "marketing", "leadership",
        "teamwork", "management", "business"
    ]

    technical_count = sum(
        keyword in skills_text
        for keyword in technical_keywords
    )

    communication_count = sum(
        keyword in skills_text
        for keyword in communication_keywords
    )

    # Convert matched skills to 1–5 scale
    programming_skill = min(5, max(1, technical_count + 2))
    communication_skill = min(5, max(1, communication_count + 2))

    # Convert to 0–100
    technical_strength = programming_skill / 5 * 100
    communication_strength = communication_skill / 5 * 100

    soft_skill_score = min(
        10,
        max(1, communication_count * 2 + 6)
    )

    return (
        programming_skill,
        communication_skill,
        soft_skill_score,
        technical_strength,
        communication_strength
    )


print("Skills mapping function created successfully!")
```

```python
# ==========================================
# CAREER EXPLANATION + SKILL GAP ANALYSIS
# ==========================================

def analyze_career_fit(student_profile, career):

    requirements = career_requirements.get(career)

    if requirements is None:
        return {
            "Career": career,
            "Why_Recommended": "Career requirement mapping not available.",
            "Skills_to_Improve": []
        }

    # Student data
    student_skills = str(student_profile.get("Skills", "")).lower()
    student_interest = str(student_profile.get("Interest", "")).lower()
    student_personality = str(student_profile.get("Personality_Type", "")).lower()

    # --------------------------------------
    # Skill matching
    # --------------------------------------

    matched_skills = []
    missing_skills = []

    for skill in requirements["skills"]:

        if skill.lower() in student_skills:
            matched_skills.append(skill)
        else:
            missing_skills.append(skill)

    # --------------------------------------
    # Interest alignment
    # --------------------------------------

    interest_match = any(
        interest.lower() in student_interest
        for interest in requirements["interests"]
    )

    # --------------------------------------
    # Explanation
    # --------------------------------------

    reasons = []

    if matched_skills:
        reasons.append(
            "Your current skills match: " +
            ", ".join(matched_skills)
        )

    if interest_match:
        reasons.append(
            "Your interest is aligned with this career."
        )

    academic = float(
        student_profile.get("Academic_Performance", 0)
    )

    aptitude = float(
        student_profile.get("Aptitude_Score", 0)
    )

    if academic >= 80:
        reasons.append(
            "Your academic performance is strong."
        )

    if aptitude >= 70:
        reasons.append(
            "Your aptitude level supports this career."
        )

    if not reasons:
        reasons.append(
            "The model identified this career as a suitable option."
        )

    return {
        "Career": career,
        "Why_Recommended": " ".join(reasons),
        "Skills_to_Improve": missing_skills
    }
```

```python
# ==========================================
# GOAL ALIGNMENT
# ==========================================

goal_career_mapping = {
    "Data Science": [
        "Data Scientist",
        "ML/AI Engineer",
        "Research Scientist",
        "Business Analyst",
        "Financial Analyst"
    ],

    "AI & Machine Learning": [
        "ML/AI Engineer",
        "Data Scientist",
        "Research Scientist",
        "Software Engineer"
    ],

    "Software Development": [
        "Software Engineer",
        "Web Developer",
        "ML/AI Engineer"
    ],

    "Cyber Security": [
        "Cyber Security Analyst"
    ],

    "Cloud Computing": [
        "Cloud Engineer",
        "Software Engineer"
    ],

    "Business Management": [
        "Business Analyst",
        "Entrepreneur",
        "Marketing Executive",
        "Sales Executive",
        "Financial Analyst",
        "HR Manager"
    ],

    "Entrepreneurship": [
        "Entrepreneur",
        "Business Analyst",
        "Marketing Executive",
        "Sales Executive"
    ],

    "Government Job": [
        "Government Officer"
    ],

    "Higher Studies": [
        "Professor",
        "Research Scientist",
        "Doctor"
    ],

    "Research": [
        "Research Scientist",
        "Data Scientist",
        "Professor"
    ]
}


def get_goal_alignment(career_goal, career):

    relevant_careers = goal_career_mapping.get(
        career_goal,
        []
    )

    if career in relevant_careers:
        # Higher alignment for direct/relevant careers
        position = relevant_careers.index(career)

        alignment_scores = [100, 90, 80, 70, 60, 50]

        if position < len(alignment_scores):
            return alignment_scores[position]

        return 50

    return 0
```

```python
# ==========================================
# FINAL CAREER SCORE
# ==========================================

def calculate_final_scores(
    student_profile,
    recommendations
):

    results = []

    for _, row in recommendations.iterrows():

        career = row["Career"]
        model_confidence = row["Confidence"]

        # Goal alignment
        goal_alignment = get_goal_alignment(
            student_profile["Career_Goal"],
            career
        )

        # Capability match
        analysis = analyze_career_fit(
            student_profile,
            career
        )

        # Number of matched skills
        matched = len([
            skill for skill in career_requirements.get(
                career, {}
            ).get("skills", [])
            if skill.lower() in str(
                student_profile["Skills"]
            ).lower()
        ])

        total_required = len(
            career_requirements.get(
                career, {}
            ).get("skills", [])
        )

        if total_required > 0:
            capability_match = (
                matched / total_required
            ) * 100
        else:
            capability_match = 50

        # Final score
        final_score = (
            0.40 * model_confidence +
            0.30 * capability_match +
            0.30 * goal_alignment
        )

        results.append({
            "Career": career,
            "Model_Confidence": round(
                model_confidence, 2
            ),
            "Capability_Match": round(
                capability_match, 2
            ),
            "Goal_Alignment": round(
                goal_alignment, 2
            ),
            "Final_Score": round(
                final_score, 2
            )
        })

    result_df = pd.DataFrame(results)

    result_df = result_df.sort_values(
        by="Final_Score",
        ascending=False
    ).reset_index(drop=True)

    result_df.insert(
        0,
        "Rank",
        range(1, len(result_df) + 1)
    )

    return result_df
```
